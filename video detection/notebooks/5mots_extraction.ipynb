```python
import sys, os
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))
from src.extration import  process_wlasl_hands, process
from src.augmentation import augment_dataset
from src.trier import tri_nom
import numpy as np
```

```python
classes = {   "thin":0,
    "go":1,
    "computer":2,
    "help":3,
    "cool":4
}
json_dir = r"../data/fichiers_json/WLASL_v0.3_cleaned.json"
mots = ["thin","go","computer","help","cool"]
wlasl_dir = r"../data/vid"
fichier_json = tri_nom(mots,json_dir)
X_wlasl,y_wlasl = process_wlasl_hands(fichier_json,wlasl_dir,100,classes)
```

```python
classes_asl = {   "THIN":0,
               "GO":1,
               "COMPUTER":2,
               "HELP":3,
               "COOL3":4
                }

asl_dir = r"../data/videos_asl"
train_csv = r"../data/train.csv"
test_csv = r"../data/test.csv"
val_csv = r"../data/val.csv"
X_asl_train,y_asl_train = process(train_csv,asl_dir,100,classes_asl)
X_asl_test,y_asl_test = process(test_csv,asl_dir,100,classes_asl)
X_asl_val,y_asl_val = process(val_csv,asl_dir,100,classes_asl)
print(f"\nX_asl_train : {X_asl_train.shape},y_asl_train : {y_asl_train.shape}")
print (f"\nX_asl_test : {X_asl_test.shape}, y_asl_test : {y_asl_test.shape}")
print(f"\n X_asl_val : {X_asl_val.shape},y_asl_val : {y_asl_val.shape}")
```

```python
X_train = np.concatenate([X_wlasl,X_asl_train],axis=0)
y_train = np.hstack([y_wlasl,y_asl_train])
print (f"\nX_train : {X_train.shape}, y_train : {y_train.shape}")
print (f"\nX_asl_val : {X_asl_val.shape}, y_asl_valtest : {y_asl_val.shape}")
print (f"\nX_asl_test : {X_asl_test.shape}, y_asl_test : {y_asl_test.shape}")
print(f"APRES ON A :")
X_train,y_train = augment_dataset(X_train,y_train)
print (f"\nX_train : {X_train.shape}, y_train : {y_train.shape}")

```

```python
np.save("../dataset_hands_only/X_train.npy",X_train)
np.save("../dataset_hands_only/X_test.npy",X_asl_test)
np.save("../dataset_hands_only/X_val.npy",X_asl_val)
np.save("../dataset_hands_only/y_train.npy",y_train)
np.save("../dataset_hands_only/y_test.npy",y_asl_test)
np.save("../dataset_hands_only/y_val.npy",y_asl_val)
```
