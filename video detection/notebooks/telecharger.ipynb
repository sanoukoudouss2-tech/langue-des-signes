```python
import json
import subprocess
import os
import zipfile

# Charger le JSON
c = r"C:/Users/sanou/Documents/Documents_1/CODE_ALPHA/Video_detection/fichiers_json/WLASL_sorted.json"
with open(c, encoding='utf-8') as f:
    data = json.load(f)

# Filtrer les video_ids qui t'intéressent (adapte le filtre ici)
video_ids = set()
for entry in data:
    for inst in entry['instances']:
       
        video_ids.add(inst['video_id'])

print(f"{len(video_ids)} vidéos à télécharger")
```

```python
out_dir = r"C:/Users/sanou/Documents/Documents_1/CODE_ALPHA/Video_detection/videos"
dataset = "risangbaskoro/wlasl-processed"
for i, vid in enumerate(video_ids, 1):
    local_mp4 = os.path.join(out_dir, f"{vid}.mp4")
    if os.path.exists(local_mp4):
        print(f"[{i}/{len(video_ids)}] {vid} déjà présent, on passe")
        continue

    remote_path = f"videos/{vid}.mp4"
    result = subprocess.run(
        ["kaggle", "datasets", "download", "-d", dataset,
         "-f", remote_path, "-p", out_dir, "--force"],
        capture_output=True, text=True
    )

    if result.returncode != 0:
        print(f"[{i}/{len(video_ids)}] Échec pour {vid}: {result.stderr.strip()}")
        continue

    # Kaggle nomme toujours le fichier "DownloadDataset" (bug connu) : on le renomme
    generic_name = os.path.join(out_dir, "DownloadDataset")
    if os.path.exists(generic_name):
        os.rename(generic_name, local_mp4)
    else:
        print(f"[{i}/{len(video_ids)}] Fichier attendu introuvable pour {vid}")
        continue

    print(f"[{i}/{len(video_ids)}] {vid} téléchargé")

print("Terminé.")
```
