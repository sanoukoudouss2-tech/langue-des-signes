```python
import sys, os
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))
from tensorflow.keras import layers,models
import numpy as np
from src.augmentation import augment_dataset
from tensorflow.keras.callbacks import EarlyStopping
early_stop = EarlyStopping(
    monitor='val_loss',       # la métrique à surveiller
    patience=15,               # nombre d'epochs à attendre sans amélioration avant d'arrêter
    restore_best_weights=True  # récupère les poids du MEILLEUR epoch, pas du dernier
)
```

```python


mod_1 = models.Sequential()
mod_1.add(layers.Input(shape=(100,132)))
mod_1.add(layers.Conv1D(64, kernel_size=5, activation='relu'))
mod_1.add(layers.MaxPooling1D(2))
mod_1.add(layers.Conv1D(64, kernel_size=5, activation='relu'))
mod_1.add(layers.LSTM(64,unroll=True))
mod_1.add(layers.Dropout(0.4))
mod_1.add(layers.Dense(5, activation='softmax'))
```

```python
mod_1.compile(optimizer='adam',
                 loss='sparse_categorical_crossentropy',
                 metrics=['accuracy'])
```

```python
X_train = np.load("../dataset_hands_only/X_train.npy")
X_test = np.load("../dataset_hands_only/X_test.npy")
X_val = np.load("../dataset_hands_only/X_val.npy")
y_train = np.load("../dataset_hands_only/y_train.npy")
y_test = np.load("../dataset_hands_only/y_test.npy")
y_val = np.load("../dataset_hands_only/y_val.npy")
```

```python
hist_1 = mod_1.fit(X_train,y_train,validation_data=(X_val,y_val),callbacks=early_stop,epochs=70,batch_size=32)
```

```python
X_inter,y_inter = augment_dataset(X_test,y_test)
X_inter2,y_inter2 = augment_dataset(X_val,y_val)
X_final = np.concatenate([X_train,X_inter,X_inter2],axis=0)
y_final = np.hstack([y_train,y_inter,y_inter2])
```

```python
mod_final = models.Sequential()
mod_final.add(layers.Input(shape=(100,132)))
mod_final.add(layers.Conv1D(64, kernel_size=5, activation='relu'))
mod_final.add(layers.MaxPooling1D(2))
mod_final.add(layers.Conv1D(64, kernel_size=5, activation='relu'))
mod_final.add(layers.LSTM(64,unroll=True))
mod_final.add(layers.Dropout(0.4))
mod_final.add(layers.Dense(5, activation='softmax'))
```

```python
mod_final.compile(optimizer='adam',
                 loss='sparse_categorical_crossentropy',
                 metrics=['accuracy'])
```

```python
hist_final = mod_final.fit(X_final,y_final,epochs=20,batch_size=32)
```

```python
import tensorflow as tf
mod_1 = tf.lite.TFLiteConverter.from_keras_model(mod_1)
mod_1 = mod_1.convert()
mod_final = tf.lite.TFLiteConverter.from_keras_model(mod_final)
mod_final = mod_final.convert()

```

```python
with open("modele_1.tflite","wb") as f:
    f.write(mod_1)
with open("modele_final.tflite","wb") as g:
    g.write(mod_final)
```
